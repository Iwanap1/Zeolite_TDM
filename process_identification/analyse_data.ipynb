```python
import json
import pandas as pd
from collections import defaultdict
import holoviews as hv

hv.extension('bokeh')

# Load JSON data
with open("../data/manual_extraction_cleaned.json", "r") as f:
    data = json.load(f)

# Step transitions, including precursor
transitions = defaultdict(int)

for doc in data:
    for sample in doc.get("samples", []):
        steps = [s["process"] for s in sample.get("post-synthesis", []) if s.get("did_step") == "true"]
        if not steps:
            continue
        # Add connection from 'precursor' to first step
        steps.insert(0, sample.get("zeolite_source").get("source"))
        transitions[("precursor", steps[0])] += 1
        # Add all internal step transitions
        for i in range(len(steps) - 1):
            transitions[(steps[i], steps[i + 1])] += 1

# Create DataFrame
df = pd.DataFrame([
    {"source": src, "target": tgt, "value": val}
    for (src, tgt), val in transitions.items()
])

# Create Chord Diagram
chord = hv.Chord(df)
chord.opts(
    labels='index',
    node_color='index',
    edge_color='source',
    cmap='Category20',
    width=700,
    height=700,
    title='Chord Diagram: Precursor to Post-Synthetic Steps'
)
```

```python
import json
import networkx as nx
import matplotlib.pyplot as plt
from collections import defaultdict
from matplotlib.lines import Line2D

# Load the JSON file
with open("../data/manual_extraction_cleaned.json", "r") as f:
    data = json.load(f)

# Initialize directed graph
G = nx.DiGraph()
node_types = {}
edge_weights = defaultdict(int)

# Build graph with weights
for doc in data:
    for sample in doc.get("samples", []):
        zeo_src = sample.get("zeolite_source").get("source")
        if isinstance(zeo_src, dict):
            source = zeo_src.get("name", "unknown")
        elif isinstance(zeo_src, str):
            source = zeo_src
        else:
            continue

        steps = [
            s["process"].replace("solvent_etching", "solution etching")
            for s in sample.get("post-synthesis", [])
        ]
        if not steps:
            continue

        # Source to first step
        edge_weights[(source, steps[0])] += 1
        node_types[source] = "source"
        node_types[steps[0]] = "step"

        # Step transitions
        for i in range(len(steps) - 1):
            edge_weights[(steps[i], steps[i + 1])] += 1
            node_types[steps[i]] = "step"
            node_types[steps[i + 1]] = "step"

# Add weighted edges to graph
for (u, v), w in edge_weights.items():
    G.add_edge(u, v, weight=w)

# Layout
pos = nx.circular_layout(G)

# Prepare nodes by type
source_nodes = [n for n, t in node_types.items() if t == "source"]
step_nodes = [n for n, t in node_types.items() if t == "step"]

# Plot
plt.figure(figsize=(11, 11))


# Scale edge thickness from 1 to 6

# Normalize edge weights for thickness scaling
max_weight = max([d['weight'] for _, _, d in G.edges(data=True)])
min_weight = min([d['weight'] for _, _, d in G.edges(data=True)])
weight_range = max_weight - min_weight if max_weight != min_weight else 1
edge_widths = [
    1 + 5 * ((d['weight'] - min_weight) / weight_range)
    for _, _, d in G.edges(data=True)
]

# Draw edges with varying thickness
node_radius = 1200  # Adjusted for better arrowhead offset
nx.draw_networkx_edges(
    G, pos,
    arrowstyle='-|>',
    arrowsize=20,
    edge_color='gray',
    width=edge_widths,
    connectionstyle='arc3,rad=0.15',
    node_size=node_radius  # <-- this offsets arrowheads before hitting nodes
)

# Nodes
nx.draw_networkx_nodes(G, pos, nodelist=source_nodes, node_color='lightcoral', node_size=1200)
nx.draw_networkx_nodes(G, pos, nodelist=step_nodes, node_color='lightblue', node_size=1200)

# Edges


# Labels
nx.draw_networkx_labels(G, pos, font_size=9)

# Edge labels

# Custom legend
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='zeolite_source',
           markerfacecolor='lightcoral', markersize=12),
    Line2D([0], [0], marker='o', color='w', label='post-synthesis',
           markerfacecolor='lightblue', markersize=12)
]
plt.legend(handles=legend_elements, loc='upper right')

plt.title("Directed Graph of Post-Synthetic Steps with Zeolite Sources", fontsize=16)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import json

# Load cleaned JSON
with open("../data/manual_extraction_cleaned.json", "r") as f:
    data = json.load(f)

# Track DOIs with calcination self-loops
calcination_self_loop_dois = set()

for entry in data:
    doi = entry.get("doi", "unknown DOI")
    for sample in entry.get("samples", []):
        steps = [s["process"] for s in sample.get("post-synthesis", []) ]
        for i in range(len(steps) - 1):
            if steps[i] == steps[i + 1] == "calcination":
                calcination_self_loop_dois.add(doi)

# Output
print("DOIs with calcination → calcination self-loops:")
for doi in sorted(calcination_self_loop_dois):
    print(doi)
```
